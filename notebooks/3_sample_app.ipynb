```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(model="gemini-3.6-flash")
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts.chat import MessagesPlaceholder
from langchain_core.prompts.chat import ChatPromptTemplate
prompt=ChatPromptTemplate.from_messages([("system","You are a food assistant, keep it short and crisp"),MessagesPlaceholder("history"), ("human","{question}"),])
chain=prompt|llm|StrOutputParser()
history=[]



```

```python
while True:
 question = input("You: ")
 if question.lower() == "quit":
   break
```

```python
answer = chain.invoke({"history": history, "question": question})   
print("AI:", answer)
history.append(("human", question))         
history.append(("ai", answer))
print(f"\n(history now has {len(history)} messages)\n")
```
